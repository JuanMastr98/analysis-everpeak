### 4.1 Calcular promedios de tráfico por ciudad

**🎯Objetivo:**
Obtener una vista consolidada del tráfico promedio por ciudad y año, para analizar patrones generales sin depender de datos diarios.

**Instrucciones**

- Agrupa los datos por `city`, `country` y `year`.
- Calcula el promedio **solo de las métricas de tráfico más relevantes**: como `jams_delay`, `traffic_index_live`, `jams_length_kms`, `jams_count`, `mins_delay`, y tiempos de viaje (`travel_time_live_per_10kms_mins` y `travel_time_hist_per_10kms_mins`).

<details>
<summary>Haz clic para ver la pista</summary>

```python
# Calcular los  promedios de trafico por ciudad, país y año
traffic_city_year_2024 =  (traffic_2024.groupby(['city', 'country', 'year'])
                          .agg({
                              'jams_delay': 'mean',
                              'traffic_index_live': 'mean',
                              'jams_length_in_kms': 'mean',
                              'jams_count': 'mean',
                              'mins_delay': 'mean',
                              'travel_time_live_per_10kms_mins': 'mean',
                              'travel_time_historic_per_10kms_mins': 'mean'
                          })
                          .reset_index())

# Mostrar resultado
traffic_city_year_2024.head()
```

### 5.1 Unir tráfico (tabla principal) con indicadores económicos

Combinar la información de tráfico y economía en un solo DataFrame para analizar cómo las condiciones económicas se relacionan con la movilidad urbana.

**Instrucciones**
- Selecciona solo las **columnas relevantes** de cada dataset (por ejemplo, variables clave de tráfico y de economía).
- Usa `.copy()` al crear subconjuntos para evitar modificar el dataset original.
- Une ambos DataFrames y define como **claves de unión** a `city` y `year`.

- Guarda el resultado en una nueva variable llamada `merged` y muestra las primeras 5 filas.


<details>

Aplica una unión de tipo "inner" para mantener las ciudades y años presentes en ambos datasets.

```python
# Seleccionar columnas clave de tráfico y economía
left_cols = ['city','country','year','jams_delay','traffic_index_live',
             'jams_length_in_kms','jams_count','mins_delay',
             'travel_time_live_per_10kms_mins','travel_time_historic_per_10kms_mins']

right_cols = ['city','year','city_gdp_capita','unemployment_percent','pm25_ug_m3','population_m']

# Usar .copy() para crear los dos nuevos datasets reducidos
# traffic_2024_small = traffic_city_year_2024[ ... ].copy()
traffic_2024_small = traffic_city_year_2024[left_cols].copy()
# eco_2024_small = ...
eco_2024_small = eco[right_cols].copy()

# Unir datasets
merged = traffic_2024_small.merge(eco_2024_small, 
                                  on=['city', 'year'], 
                                  how='inner')

# Mostrar las primeras 5 filas
print(merged.head())
```

```python
# Crear boxplot para observar el comportamiento de los minutos de congestion JamsDelay
# crea tu gráfico

plt.boxplot(merged['jams_delay'], vert=True)
plt.ylabel('Retraso por Congestión (segundos)', fontsize=12)
plt.xlabel('JamsDelay', fontsize=12)
# obtener promedio para mostrarlo en título
mean_value = merged['jams_delay'].mean()
plt.title(f'Boxplot de JamsDelay (2024)\nPromedio: {mean_value:.2f}')
plt.show()
```

```python
# Crear histograma para ver la distribución de la economía (city_gdp_capita)


plt.hist(merged['city_gdp_capita'])

mean_gdp = merged['city_gdp_capita'].mean()
plt.axvline(mean_gdp, color='red', linestyle='--', linewidth=2, label=f'Promedio: ${mean_gdp:.2f}')

plt.xlabel('PIB per Cápita (USD)', fontsize=12)
plt.ylabel('Frecuencia', fontsize=12)
plt.title(f'Histograma de Distribución del PIB per Cápita (2024)\nPromedio: ${mean_gdp:.2f}', 
          fontsize=14, fontweight='bold')
plt.legend(fontsize=11)
plt.grid(axis='y', alpha=0.3)
plt.show()
```

```python
# Gráfico de barras para comparar jams_delay y city_gdp_capita por ciudad
#merged.plot( ... , y=['jams_delay', 'city_gdp_capita'])
city_comparison = merged.groupby('city')[['jams_delay', 'city_gdp_capita']].mean()
city_comparison_normalized = city_comparison.copy()
city_comparison_normalized['jams_delay'] = (city_comparison['jams_delay'] - city_comparison['jams_delay'].min()) / (city_comparison['jams_delay'].max() - city_comparison['jams_delay'].min())
city_comparison_normalized['city_gdp_capita'] = (city_comparison['city_gdp_capita'] - city_comparison['city_gdp_capita'].min()) / (city_comparison['city_gdp_capita'].max() - city_comparison['city_gdp_capita'].min())

fig, ax = plt.subplots(figsize=(14, 6))

x = np.arange(len(city_comparison_normalized))
width = 0.35

bars1 = ax.bar(x - width/2, city_comparison_normalized['jams_delay'], width, 
               label='JamsDelay (Normalizado)', color='skyblue', edgecolor='black')
bars2 = ax.bar(x + width/2, city_comparison_normalized['city_gdp_capita'], width, 
               label='PIB per Cápita (Normalizado)', color='lightcoral', edgecolor='black')

ax.set_xlabel('Ciudad', fontsize=12)
ax.set_ylabel('Valor Normalizado (0-1)', fontsize=12)
ax.set_title('Comparación de JamsDelay vs PIB per Cápita por Ciudad (2024)', 
             fontsize=14, fontweight='bold')
ax.set_xticks(x)
ax.set_xticklabels(city_comparison_normalized.index, rotation=45, ha='right')
ax.legend(fontsize=11)
ax.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.xticks(rotation=90)
plt.show()
```

**Tip:** Antes del `plt.show()` agrega el código `plt.xticks(rotation=90)` para rotar las etiquetas del eje X en 90 grados.

CONTEXTO Y OBJETIVO

Este análisis examina la relación fundamental entre la movilidad urbana (congestión vehicular y tiempos de viaje) y la productividad económica (PIB per cápita) en ciudades latinoamericanas durante 2024. ¿Una ciudad con alta congestión vehicular puede mantener elevada productividad económica, o existe una relación inversa que debilita la competitividad?

Las variables clave analizadas incluyen: Jams Delay (retraso por congestión en segundos), Traffic Index Live (índice de congestión en vivo), Travel Time (tiempo de viaje normalizado), y PIB per Cápita (indicador de productividad). Estas métricas son relevantes para la toma de decisiones porque permiten identificar si invertir en infraestructura de transporte generaría retorno económico directo, al reducir cuellos de botella que limitan la productividad laboral y empresarial.

COBERTURA DE DATOS Y METODOLOGÍA

El análisis integra datos de 2024 de 12 ciudades distribuidas en 4 países latinoamericanos (Argentina, Brasil, Colombia, Perú). La base combinada contiene 30 registros tras la unión de datos de tráfico (TomTom) y datos económicos (OECD).
